```python
import matplotlib.pyplot as plt
import math
from PIL import Image
from matplotlib.patches import Patch


from matplotlib.gridspec import GridSpec
def plot_final_output(img_dict):
    plt.rcParams['figure.dpi'] = 300
    plt.rcParams['savefig.dpi'] = 300
    cols = 2
    rows = math.ceil(len(img_dict)/cols)

    fig = plt.figure(figsize=(10, rows*3))
    gs = GridSpec(rows, cols, figure=fig, wspace=-0.15, hspace=0.0)

    # Column titles positioned above columns
    fig.text(0.19, 0.98, 'XGBoost', fontsize=10, ha='center', va='bottom', weight='bold')
    fig.text(0.57, 0.98, 'Random Forest', fontsize=10, ha='center', va='bottom', weight='bold')

    for idx, (key, path) in enumerate(img_dict.items()):
        row = idx // cols
        col = idx % cols
        ax = fig.add_subplot(gs[row, col])
        img = Image.open(path)
        ax.imshow(img)
        ax.axis('off')

    # Legend: green=unburned, red=burned (top right)
    legend_elements = [Patch(facecolor='#b6f2c0', label='Unburned'),
                    Patch(facecolor='#8B0000', label='Burned')]

    fig.legend(handles=legend_elements, loc='upper right',fontsize=8, bbox_to_anchor=(0.875, 0.999))


    plt.subplots_adjust(top=0.98) 
    plt.show()

    # Save high-quality version
    plt.savefig('MachineLearning/output/prediction_img/combined_single.png', dpi=300, bbox_inches='tight', facecolor='white')
```

### Result plot

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
schemes_single = ['#1', '#2', '#3', '#4']
xg_single = [0.7434, 0.7568, 0.8045, 0.8102]
rf_single = [0.7686, 0.7776, 0.8124, 0.8143]

schemes_multi = ['#5', '#6', '#7', '#8']
xg_multi = [0.7875, 0.7854, 0.8272, 0.8418]
rf_multi = [0.8086, 0.8209, 0.8443, 0.8515]

x = np.arange(len(schemes_single))
width = 0.35

# --- Single-Date Plot ---
plt.figure(figsize=(5,3)) 
bars1 = plt.bar(x - width/2, xg_single, width, label='XGBoost')
bars2 = plt.bar(x + width/2, rf_single, width, label='Random Forest')

# Add labels on bars
for bar in bars1:
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01, 
             f'{bar.get_height():.3f}', ha='center', va='bottom', fontsize=7)
for bar in bars2:
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01, 
             f'{bar.get_height():.3f}', ha='center', va='bottom', fontsize=7)

plt.xticks(x, schemes_single)
plt.xlabel('Schemes')
plt.ylabel('F1-Score')
plt.title('Single-Date Comparison')
plt.ylim(0,1)
plt.legend(loc='upper left',fontsize=6,bbox_to_anchor=(0.76, 1.2))
plt.tight_layout()
plt.show()

# --- Multi-Date Plot ---
x2 = np.arange(len(schemes_multi))

plt.figure(figsize=(5,3)) 
bars1 = plt.bar(x2 - width/2, xg_multi, width, label='XGBoost', )
bars2 = plt.bar(x2 + width/2, rf_multi, width, label='Random Forest')

# Add labels on bars
for bar in bars1:
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01, 
             f'{bar.get_height():.3f}', ha='center', va='bottom', fontsize=7)
for bar in bars2:
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01, 
             f'{bar.get_height():.3f}', ha='center', va='bottom', fontsize=7)

plt.xticks(x2, schemes_multi)
plt.xlabel('Schemes')
plt.ylabel('F1-Score')
plt.title('Multi-Date Comparison')
plt.ylim(0,1)
plt.legend(loc='upper left',fontsize=6,bbox_to_anchor=(0.76, 1.2))
plt.tight_layout()
plt.show()
```
